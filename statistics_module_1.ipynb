Q1.  What is the difference between descriptive statistics and inferential
      statistics? Explain with examples.

ans-  Descriptive statistics are used to summarize and describe the characteristics of a data set. They provide a concise overview of the data, allowing for easy interpretation of its main features. They do not involve generalizing beyond the data at hand.
Example: Measures of Central Tendency: Mean, median, and mode describe the center of the data.

Inferential statistics use data from a sample to make inferences or predictions about a larger population. The goal is to draw conclusions that extend beyond the immediate data.

Example:
Hypothesis Testing: Used to test a claim or hypothesis about a population parameter based on sample data.
  
Q2. What is sampling in statistics? Explain the differences between random
and stratified sampling.

ans-  Sampling is a process used in statistical analysis in which a predetermined number of observations are taken from a larger population. The goal of sampling is to select a representative subset of the population to draw conclusions about the entire population without having to collect data from every single member.

Random Sampling
Random sampling is a basic sampling technique where every member of the population has an equal and independent chance of being selected for the sample. This method aims to ensure that the sample is representative of the population, reducing bias.

Stratified sampling is a method that involves dividing a population into smaller subgroups called 'strata' based on shared characteristics (e.g., age, gender, income level, educational attainment). After dividing the population into strata, a random sample is then drawn from each stratum. This ensures that each subgroup is adequately represented in the overall sample.

Q3. Define mean, median, and mode. Explain why these measures of central
tendency are important.

The mean is calculated by summing all the values in a dataset and then dividing by the number of values in that dataset.

Median
The median is the middle value in a dataset when the values are arranged in ascending or descending order. If the dataset has an odd number of values, the median is the single middle value. If the dataset has an even number of values, the median is the average of the two middle values.

Mode
The mode is the value that appears most frequently in a dataset. A dataset can have one mode (unimodal), multiple modes (multimodal), or no mode if all values appear with the same frequency.

Q4. Explain skewness and kurtosis. What does a positive skew imply about the
data?

ans-  Skewness
Skewness is a measure of the asymmetry of the probability distribution of a real-valued random variable about its mean.

Positive Skew (Right Skew): A distribution is positively skewed if its tail is longer on the right side. This means that there are more extreme values on the higher end of the distribution. In a positively skewed distribution, the mean is typically greater than the median, which is often greater than the mode.

Kurtosis
Kurtosis is a measure of the "tailedness" of the probability distribution of a real-valued random variable. In other words, it describes the shape of the distribution's tails in relation to its central peak. It indicates how much of the variance comes from infrequent extreme deviations (tails) versus frequent modestly sized deviations (shoulders).

Q6. Compute the covariance and correlation coefficient between the following two datasets provided as lists in Python: list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

ans-  ###import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]


array_x = np.array(list_x)
array_y = np.array(list_y)

covariance_matrix = np.cov(array_x, array_y)
covariance = covariance_matrix[0, 1] # The covariance between x and y

correlation_coefficient = np.corrcoef(array_x, array_y)[0, 1]

print(f"List X: {list_x}")
print(f"List Y: {list_y}")
print(f"Covariance between X and Y: {covariance}")
print(f"Correlation coefficient between X and Y: {correlation_coefficient}") ###

Q7. Write a Python script to draw a boxplot for the following numeric list and
identify its outliers. Explain the result:
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

ans.  ###import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

plt.figure(figsize=(8, 6))
sns.boxplot(y=data, color='skyblue')
plt.title('Boxplot of the Numeric List')
plt.ylabel('Values')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1


lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR


outliers = [x for x in data if x < lower_bound or x > upper_bound]

print(f"Data: {data}")
print(f"Q1 (25th percentile): {Q1}")
print(f"Q3 (75th percentile): {Q3}")
print(f"IQR (Interquartile Range): {IQR}")
print(f"Lower Bound for Outliers: {lower_bound}")
print(f"Upper Bound for Outliers: {upper_bound}")
print(f"Identified Outliers: {outliers}")

Q9. Your team has collected customer satisfaction survey data on a scale of
1-10 and wants to understand its distribution before launching a new product.
● Explain which summary statistics and visualizations (e.g. mean, standard
deviation, histogram) you’d use.
● Write Python code to create a histogram using Matplotlib for the survey data:
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

ans-  Histogram: This is a crucial visualization for understanding the distribution of the data. It shows the frequency of each score (or ranges of scores) and helps to identify:

Shape: Is the distribution symmetric, skewed (left or right), or bimodal?
Central Tendency: Where are most of the scores concentrated?
Spread: How wide is the range of scores?
Outliers/Unusual Values: Are there any scores far removed from the majority

###
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]


mean_score = np.mean(survey_scores)
median_score = np.median(survey_scores)
std_dev_score = np.std(survey_scores)
min_score = np.min(survey_scores)
max_score = np.max(survey_scores)

print(f"Survey Scores: {survey_scores}")
print(f"Mean Score: {mean_score:.2f}")
print(f"Median Score: {median_score:.2f}")
print(f"Standard Deviation: {std_dev_score:.2f}")
print(f"Minimum Score: {min_score}")
print(f"Maximum Score: {max_score}")


plt.figure(figsize=(10, 6))
sns.histplot(survey_scores, bins=np.arange(min(survey_scores), max(survey_scores) + 2) - 0.5, kde=False, color='skyblue', edgecolor='black')
plt.title('Distribution of Customer Satisfaction Survey Scores')
plt.xlabel('Satisfaction Score (1-10)')
plt.ylabel('Frequency')
plt.xticks(range(min(survey_scores), max(survey_scores) + 1)) # Ensure integer ticks for scores
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
